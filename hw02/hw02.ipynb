````python
# 1
def to_nums(s):
    return float(str(s).replace(" ",""))
for i in rows:
    i['amount'] = to_nums(i['amount'])


````

````python
def clean_city(s):
    s = s.strip().title()
    if s == "Shymkent":
        return "Шымкент"
    elif s == "Astana":
        return "Астана"
    elif s == "Almaty":
        return "Алматы"
    else: 
        return s

for i in rows:
    i['city'] = clean_city(i['city'])
    
````

````python
print({i['city'] for i in rows})
````

````python
# 2
categ_count = {}
for i in rows:
    if i['category'] in categ_count:
        categ_count[i['category']] = categ_count.get(i['category'],0) + 1
    else:
        categ_count[i['category']] = 1
````

````python
sorted(categ_count, key=categ_count.get, reverse=True)
````

````python
# 3 
print({row['status'] for row in rows})
````

````python
revenue = {}
orders = {}

for row in rows:
    if row['status'] == 'оплачен':
        cat = row['category']
        revenue[cat] = revenue.get(cat, 0) + row['amount']
        orders[cat] = orders.get(cat, 0) + 1
````

````python
print(f"{'Категория':<12}|{'Выручка':>11}|{'Ср. чек':>7}")
for cat in revenue:
    avg = round(revenue[cat] / orders[cat])
    print(f"{cat:<12}|{revenue[cat]:>11}|{avg:>3}")
````

````python
# 4
month_count = {}
for row in rows:
    month = row['order_date'][:7]
    month_count[month] = month_count.get(month, 0) + 1

print(month_count)
````
