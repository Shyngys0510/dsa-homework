````python
# База
cities_clean = [i.strip().title() for i in cities]
print(cities_clean)
````

````python
amounts_int = [int(i) for i in amounts]
print(amounts)
print(amounts_int)
````

````python
sum_amount = sum(amounts_int)
len_amount = len(amounts_int)
avg_check = round(sum_amount / len_amount)
print(f"Общая сумма:{sum_amount}; Число заказов:{len_amount}; Средний чек:{avg_check}.")
````

````python
payment = list(zip(amounts_int,paid))
paids = []
for i in payment:
    if i[1] == False:
        paids.append(i)
unpaid = sum(i[0] for i in paids)
print(F"Сумма не оплаченных:{unpaid}, Кол-во не оплаченных:{len(paids)}")
````
