````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````


````python
# Стандарт
````

````python
x = amount[(amount>50000) & (paid == True)]
print(f"До маркетинга - Выручка:{x.sum()},Количество заказов:{len(x)},Средний чек:{x.mean()}")
y = amount[(amount>50000) & (paid == True)] * 0.85
print(f"После маркетинга - Выручка:{y.sum()},Количество заказов:{len(y)},Средний чек:{y.mean()}")
print(F"Потеря выручки магазина с внедрение маркетинга:{x.sum() - y.sum()}")
````

````python
x = amount[(amount>50000) & (paid == True) & (cust != 'TEST')]
print(f"До маркетинга - Выручка:{x.sum()},Количество заказов:{len(x)},Средний чек:{x.mean()}")
y = amount[(amount>50000) & (paid == True) & (cust != 'TEST')] * 0.85
print(f"После маркетинга - Выручка:{y.sum()},Количество заказов:{len(y)},Средний чек:{y.mean()}")
print(F"Потеря выручки магазина с внедрение маркетинга:{x.sum() - y.sum()}(Без заказов клиентов TEST)")
````

#### Различие суммы потерь со всеми клиентами и без клиентов тест заключаются в средних значениях, но так как мы видим что количество заказов уменьшилась всего на 3 продажи вероятнее всего в этих 3 продажах есть выбросы то есть аномально высокие значения, так как всего из за 3 этих значениях среднйи чек падает аж в 4,5 раза.
